```python
import pandas as pd

df = pd.DataFrame({
    "emp_id":[1,2,3,4,5],
    "dept":["IT","HR","IT","Sales","HR"],
    "salary":[50000,40000,None,70000,45000],
    "age":[25,30,28,35,29]
})

df
```

```python
# Q1

# Find employees with salary > 45000.

# Follow-up:

# Use loc
# Use query

df[ df['salary'] > 45000 ]
```

```python
df.loc[df['salary'] > 45000, ['emp_id', 'dept'] ] 
```

```python
df.query("salary > 45000")
```

```python
# Q2

# Find missing values.

# Follow-up:

# Fill with department-wise average salary.


df.isnull()
```

```python
dept_wise_avg_sal = df.groupby(['dept'])['salary'].mean()
dept_wise_avg_sal
```

```python
# Follow-up: Fill with department-wise average salary
df['salary'] = df.groupby('dept')['salary'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
# Q3 ⭐

# Department-wise average salary.

df.groupby('dept')['salary'].mean()
```

```python
# Q4 ⭐⭐

# Find highest-paid employee in each department.

# df.groupby('dept')['salary'].idxmax()
df.loc[ df.groupby('dept')['salary'].idxmax() ]
```

```python
df.groupby('dept').apply(lambda x: x.nlargest(1, "salary"))
```

```python
# Q5 ⭐⭐⭐

# Add column:

# salary_category

# Rules:

# <50000 -> Low
# 50000-60000 -> Medium
# >60000 -> High
```

```python
df['salary_category'] = df['salary'].apply(lambda x: "Low" if x < 50000 else   "Medium" if (x >= 50000 and x <60000) else "High")
df
```

```python
pd.cut(df['salary'], 
    bins=[0, 50000, 60000, float("inf")],
    labels=['Low', "Medium", "High"])
```

```python
# Q6 ⭐⭐⭐

# Create:

# dept_summary

# containing:

# dept
# employee_count
# avg_salary
# max_salary

df.groupby('dept').agg(employee_count=("emp_id", "size"), avg_salary=("salary", "mean"), max_salary=("salary", "max"))
```

```python
import pandas as pd

customers = pd.DataFrame({
    "cust_id":[101,102,103,104,105,106],
    "customer_name":["Aman","Nikhil","Karan","Rahul","Priya","Neha"],
    "city":["Delhi","Mumbai","Delhi","Pune","Mumbai","Bangalore"]
})

orders = pd.DataFrame({
    "order_id":[1,2,3,4,5,6,7,8],
    "cust_id":[101,102,101,103,107,104,102,108],
    "amount":[500,1200,800,300,1500,700,1000,900],
    "category":["Electronics","Fashion","Fashion","Grocery",
                "Electronics","Fashion","Grocery","Fashion"]
})
```

```python
# Q1 (Warm-up)

# Merge both tables.

df = customers.copy()

df.merge(orders, on="cust_id", how="outer")
```

```python
# Q2 (Real Business Question)

# Find orders whose customer information is missing.

df.merge(orders, on="cust_id", how="right").query("customer_name.isnull()")
```

```python
merged = df.merge(orders, on="cust_id", how="right")
missing_info = merged[merged["customer_name"].isnull()]
missing_info
```

```python
# Q3 (Frequently Asked)

# Find customers who never placed any order.

df.merge(orders, on='cust_id', how='left').query("order_id.isnull()")
```

```python
merged_df = df.merge(orders, on="cust_id", how="left")
merged_df[ merged_df['order_id'].isnull() ]
```

```python
# Q4 (Product Company Level)

# City-wise analysis:

# Return:

# city	total_orders	total_revenue

merged_df = df.merge(orders, on='cust_id', how="outer")
merged_df
```

```python
merged_df.groupby('city').agg(total_orders = ("order_id", "count"), revenue=("amount", "sum"))
```

```python
# Q5 (Good 5-Year Question)

# Category-wise:

# Total Revenue
# Average Revenue
# Number of Orders

merged_df.groupby("category").agg(revenue=("amount", "sum"), avg_revenue=("amount", "mean"), number_of_orders=("order_id", "count"))
```

```python
# Q6 (Very Common)

# Find Top 3 customers by revenue.

# Expected output:

# cust_id	revenue
# Follow-up

# Return customer name as well.

result = merged_df.groupby('cust_id')['amount'].sum().reset_index()[:3]
result
```

```python
# result.cust_id.isin(customers['cust_id'])

result.merge(customers, on="cust_id", how="left")[['customer_name', 'amount', 'cust_id']]
```

```python
# Q7 (Strong Candidate Question)

# For each city:

# Find customer who generated maximum revenue.

# (No loops)


# simple and easy soln
merged_df.groupby(['city', 'cust_id'])['amount'].sum().reset_index()\
.sort_values(["city","amount"], ascending=[True,False])\
.groupby('city').head(1)
```

```python
# For each city: customer with max revenue
max_rev = (
    merged_df.groupby(["city","cust_id"])["amount"].sum().reset_index()
)

result = max_rev.loc[max_rev.groupby("city")["amount"].idxmax()]
result
```

```python
# Q8 (Interview Twist)

# Suppose:

orders["amount"] = [500,None,800,300,None,700,1000,900]

# Now:

# Calculate city revenue.
# Ignore missing values properly.
# Follow-up

# Should you use:

# mean
# median
# zero
# for imputation?

# Why?

result = customers.merge(orders, on='cust_id', how="outer")
```
