```python
ls = [ 14, 18, 20, [20, 30, 40], 15 ]
ls
```

```python
# its a new list var
new_ls = []

# this for loop
for i in ls:
    if type(i)==list:
        new_ls.extend(i)
    else:
        new_ls.append(i)

```

```python
new_ls
```

```python
dict_ = {}

for i in new_ls:
    if i in dict_:
        dict_[i] = dict_[i]+1
    else:
        dict_[i] = 1

dict_
```

```python
for k in dict_.keys():
    if dict_[k] > 1:
        print(k)
    
```
