```python
import numpy as np

txn = np.array([500, 200, 800, 200, 1000, 500, 300])
```

```python
# Part A — where()

# Find positions where transaction amount is greater than 500.

# Expected:
# [2, 4]

idx = np.where(txn > 500)
txn[idx]

# Follow-up:
# Replace all transactions > 500 with 999.
txn[txn >500] = 999
```

```python
txn
```

```python
# Part B — argmax()

# Without finding the value directly:

txn = np.array([500, 200, 800, 200, 1000, 500, 300])

# Find the index of the highest transaction.

print(np.argmax(txn))
print()

# Follow-up:
# How would you get the actual value using that index?
print(txn[np.argmax(txn)])
```

```python
# Part C — argsort() (Very Common)

# Find indices that would sort the array.

# Expected concept:
# np.argsort(txn)
sorted_txn = np.argsort(txn)
print(sorted_txn)
print(txn[sorted_txn])



# Follow-up:

# Get:

# Top 1 transaction
print(txn[sorted_txn][0])

# Top 3 transactions
print(txn[sorted_txn][:3])

# Top 3 transaction indices
print(sorted_txn[:3])

# Without using loops.
```

```python
# Part D — unique()

# Find all unique transaction values.

# Expected:
# [200, 300, 500, 800, 1000]

txn
```

```python
np.unique( txn )
```

```python
# Part E — isin() (Production Use Case)

# Suppose fraud transactions are:
fraud = [200, 1000]

# Return only fraud transactions from txn.
# Expected:
# [200, 200, 1000]

txn [ np.isin(txn, fraud) ]
```

```python
# non fraud transactions

txn [~ np.isin(txn, fraud) ]
```

```python
# 6. Case Study: Customer Purchase Matrix
```

```python
import numpy as np

sales = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900],
    [1000, 1100, 1200]
])

sales

# Rows = Customers
# Columns = Product Categories
```

```python
# Part A — Reshape vs Resize
# Q1

# Convert:

# (4,3)

# to

# (2,6)

# flatteing the numpy array
print(sales.reshape(-1))

print()

# chnaging shape of np array
print( sales.reshape(2,6))
```

```python
# reshape -> create a new array with new shape (like view)

# resize -> changes shape or size of original array
```

```python
# Part B — Flatten vs Ravel
# Q2

sales = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900],
    [1000, 1100, 1200]
])

# Convert:

# (4,3)

# to

# (12,)

# flatening the array -> flatten -> create new array of flat dim
print(sales.flatten())
print()

# print(sales.ravel())
# print()

# sales[2] = 780
# print(sales)

# sales[2][1] = 890
# print(sales)
```

```python
# flatten -> create a deep copy; changes not reflect in original array, slow and memory consuming!

# ravel -> creates a reference (view) in memory; faster and efficient; changes not reflect in orginal array

```

```python
# Part C — Transpose
# Q3

# Convert:

# (4,3)
print(sales)
print(sales.shape)
print()
# to

# (3,4)

print(sales.T)
print(sales.T.shape)
```

```python
# Part - D: Matrix Multiplications

weights = np.array([
    [0.2],
    [0.3],
    [0.5]
])

print(weights)

print()

print(sales.T * weights)
```

```python
np.dot(sales, weights)
```

```python
sales @ weights
```

| Function                                  | Purpose                                         |
| ----------------------------------------- | ----------------------------------------------- |
| `np.random.seed(42)`                      | Fix randomness (reproducibility)                |
| `np.random.rand(5)`                       | Random floats between 0 and 1                   |
| `np.random.randn(5)`                      | Random values from standard normal distribution |
| `np.random.randint(10, 100, 5)`           | Random integers                                 |
| `np.random.uniform(1000, 5000, 5)`        | Random floats in a range                        |
| `np.random.normal(50000, 10000, 5)`       | Normal distribution (mean, std)                 |
| `np.random.choice([1,2,3], 5)`            | Random selection from values                    |
| `np.random.choice(arr, 5, replace=False)` | Random sample without replacement               |
| `np.random.shuffle(arr)`                  | Shuffle original array in-place                 |
| `np.random.permutation(arr)`              | Shuffled copy                                   |
| `np.random.random(5)`                     | Same as `rand()`, floats 0–1                    |
| `np.random.bytes(10)`                     | Random bytes (rarely used in DS)                |


Every time you run it, you'll get the same random numbers.

np.random.seed() makes random results reproducible, which is important for debugging, ML experiments, and model comparisons.

shuffle()      -> modifies original

permutation()  -> returns new shuffled array
