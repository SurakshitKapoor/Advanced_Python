```python
states = np.random.choice(['DL', 'HR', 'MH', 'TN'], size=12)
age = np.random.randint(18, 29, size=12)
marks = np.round(np.random.rand(12)*1000, 3)
# marks
marital_status = np.random.choice((True, False), size=(12))

days = np.random.randint(1, 30, size=(12))
# months = np.random.choice(['Jan', 'Feb', "Mar", 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec'], size=12)
months = np.random.randint(1, 13, size=12)
years = np.random.choice([2024, 2025, 2026], size=12)

emp_id = [ "emp_"+str(np.random.randint(150, 500, 1)[0]) for i in range(0, 12) ]
emp_id
```

```python
# try -1 : to get dates!
(f"{days}-{months}-{years}")
```

```python
# try-2 : to get dates
# str(days)+"/"+months+"/"+str(years)
```

```python
# try-3 : to get dates using loop -> success

dates = np.array([ str(days[i])+"-"+str(months[i])+"-"+str(years[i]) for i in range(0, len(days)) ])
print(type(dates))
print()
print(dates)
print()
```

```python
dates.dtype
```

```python
# arr_dt = dates.astype("datetime64[D]")
```

```python
# converting dates to proper type using pandas and converting back to numpy!

dates = pd.to_datetime(dates, dayfirst=True).to_numpy(dtype="datetime64[D]")
```

```python
dates
```

```python
# create a dataframe from all individual sources!

df = pd.DataFrame({
    "emp_id":emp_id,
    "states":states,
    "marks":marks,
    "marital_status":marital_status,
    "joining_dates":dates
})
df
```

```python
df.info()
```

```python
# extracting detailes from date cols.

df['joining_month'] = df['joining_dates'].dt.month_name()
df['joining_year'] = df['joining_dates'].dt.year

df['joining_day_of_week'] = df['joining_dates'].dt.dayofweek
df['joining_day_name'] = df['joining_dates'].dt.day_name()

# np.where -> to get indexes of elements based on condition, to use as conditional statement as well!
# df['joining_day_type'] = np.where(
#     df['joining_dates'].dt.dayofweek >= 5, "Weekend", "Weekday"
# )

# df[['joining_dates']].applymap(lambda x: "Weekend" if x.dayofweek >= 5 else "Weekday" )
# Correct way → apply on Series
df['joining_day_type'] = df['joining_dates'].apply(
    lambda x: "Weekend" if x.dayofweek >= 5 else "Weekday"
)
```

```python
df
```

```python
# in marks, break on basis of . and check does left ad right part are above 200 , if yes count those!

# conversion -> split -> divide into cols. -> type conversion -> comparison -> sum (row-wise) -> list conversion
(df['marks'].astype(str).str.split(".", expand=True).astype(float) > 200).sum(axis=1).to_list()
```
